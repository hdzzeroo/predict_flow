```python
from typing_extensions import TypedDict
from langgraph.graph import StateGraph

# 定义状态结构
class TrafficState(TypedDict):
    foo: str

# 定义节点
def chatbot(state: TrafficState) -> dict: return {}
def visualization(state: TrafficState) -> dict: return {}
def cluster(state: TrafficState) -> dict: return {}
def draw_hulls(state: TrafficState) -> dict: return {}
def report(state: TrafficState) -> dict: return {}

# 初始化工作流
builder = StateGraph(state_schema=TrafficState)

# 添加节点和边
builder.add_node("chatbot", chatbot)
builder.add_node("visualization", visualization)
builder.add_node("cluster", cluster)
builder.add_node("draw_hulls", draw_hulls)
builder.add_node("report", report)

builder.set_entry_point("chatbot")
builder.add_edge("chatbot", "visualization")
builder.add_edge("visualization", "cluster")
builder.add_edge("cluster", "draw_hulls")
builder.add_edge("draw_hulls", "report")
builder.set_finish_point("report")

# 编译工作流
compiled = builder.compile()

# 获取并输出 SVG 图像
svg = compiled.get_graph().draw_svg()
with open("langgraph_workflow.svg", "w") as f:
    f.write(svg)

print("✅ 流程图已生成为 langgraph_workflow.svg")
```
