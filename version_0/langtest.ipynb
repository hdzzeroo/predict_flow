```python
"""
traffic_workflow.py
LangGraph 框架：模拟人类交通预测流程

⚠️ 仅提供整体骨架，每个节点的业务逻辑请按需补充。
"""

from typing import TypedDict, List, Dict, Any

# LangGraph 核心
from langgraph.graph import StateGraph
```

```python


# ────────────────────────────────────────────────────────────
# 1. 定义全局状态
# ────────────────────────────────────────────────────────────

class TrafficState(TypedDict, total=False):
    # —— 用户输入 ——
    user_input: str                     # 原始指令
    # —— Chatbot 节点输出 ——
    file_path: str                      # CSV 路径
    route: str                          # 路段标识（可选）
    ts: str                             # 时间字符串（可选）
    # —— 可视化节点输出 ——
    triangles: List[Dict[str, Any]]     # 小三角形数据
    fig_path: str                       # 图片文件路径
    # —— 聚类节点输出 ——
    clusters: List[List[int]]           # 三角形索引簇
    # —— 大三角形节点输出 ——
    hulls: List[Dict[str, Any]]         # 外包大三角形
    # —— 报告节点输出（终点） ——
    report: str                         # LLM 生成的分析报告


# ────────────────────────────────────────────────────────────
# 2. 各节点占位实现（返回新的部分状态）
# ────────────────────────────────────────────────────────────


def chatbot_node(state: TrafficState) -> TrafficState:
    """解析 user_input，提取路名 / 时间 → 生成 file_path 等。
    这里暂时使用占位返回。"""
    return {
        **state,
        "file_path": "<待实现>",     # ➜ 请在此拼接真实路径
        "route": state.get("route", ""),
        "ts": state.get("ts", ""),
    }


def load_visualize_node(state: TrafficState) -> TrafficState:
    """读取 CSV，绘制小三角形，可视化。"""
    return {
        **state,
        "triangles": [],               # ➜ 填充三角形坐标数据
        "fig_path": "<待实现>",
    }


def cluster_triangles_node(state: TrafficState) -> TrafficState:
    """对 triangles 做聚类，输出 clusters。"""
    return {
        **state,
        "clusters": [],                # ➜ 聚类结果
    }


def draw_hulls_node(state: TrafficState) -> TrafficState:
    """为每个簇绘制外包大三角形，可更新 fig_path。"""
    return {
        **state,
        "hulls": [],                   # ➜ 大三角形坐标
        # 如需更新可视化文件，则替换 fig_path
    }


def llm_report_node(state: TrafficState) -> TrafficState:
    """根据 hulls 调用 LLM 生成报告。"""
    return {
        **state,
        "report": "<待实现>",
    }


# ────────────────────────────────────────────────────────────
# 3. 构建 LangGraph 图
# ────────────────────────────────────────────────────────────

def build_workflow():
    """组装并返回可调用的 LangGraph workflow。"""
    g = StateGraph(TrafficState)

    # 注册节点
    g.add_node("chatbot", chatbot_node)
    g.add_node("load_visualize", load_visualize_node)
    g.add_node("cluster_triangles", cluster_triangles_node)
    g.add_node("draw_hulls", draw_hulls_node)
    g.add_node("llm_report", llm_report_node)

    # 串联边
    g.set_entry_point("chatbot")
    g.connect("chatbot", "load_visualize")
    g.connect("load_visualize", "cluster_triangles")
    g.connect("cluster_triangles", "draw_hulls")
    g.connect("draw_hulls", "llm_report")
    g.set_final_node("llm_report")

    # 编译
    return g.compile()


# ────────────────────────────────────────────────────────────
# 4. 简单测试
# ────────────────────────────────────────────────────────────

if __name__ == "__main__":
    workflow = build_workflow()
    # 示例调用
    result = workflow.invoke({
        "user_input": "请分析6月3日东名高速东京⇄横滨区间的交通情况"
    })
    print("最终报告\n", result.get("report"))
```
